```python
#Task 1: Multiply two 2x2 matrices of Instagram likes and comments
# Rows = post 1, post 2 | Columns = likes, comments (for day 1)
day1 = [
    [120, 15],
    [200, 30],
]

# Day 2 values used as the second matrix
day2 = [
    [2, 1],
    [1, 2],
]

def multiply_matrices(a, b):
    result = [[0, 0], [0, 0]]
    for i in range(2):
        for j in range(2):
            total = 0
            for k in range(2):
                total += a[i][k] * b[k][j]
            result[i][j] = total
    return result


result = multiply_matrices(day1, day2)

print("Resulting matrix:")
for row in result:
    print(row)
```

```python
#Task 2: Linear regression prediction function
def linear_regression_predict(x, m, c):
    return m * x + c


# quick test
print(linear_regression_predict(5, 2, 3))
```

```python
#Task 3: Predict Zomato delivery time based on distance
distances = [2, 5, 8]   # in km
m = 4
c = 10

for d in distances:
    predicted_time = linear_regression_predict(d, m, c)
    print("Distance:", d, "km -> Predicted delivery time:", predicted_time, "minutes")
```
