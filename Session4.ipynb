```python
#Task 1: Plot daily Swiggy delivery orders and check skewness
import matplotlib.pyplot as plt

orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]
days = list(range(1, 11))

plt.bar(days, orders)
plt.xlabel("Day")
plt.ylabel("Orders")
plt.title("Daily Orders")
plt.show()
```

```python
# Most days the orders are close together (around 12-18), but day 7 has a
# huge spike of 200 orders. This one high value pulls the data toward the
# right, so this data is right skewed.
```

```python
#Task 2: Mean vs Median of Instagram follower counts
followers = [5, 7, 8, 9, 10, 12, 15, 95]

mean_val = sum(followers) / len(followers)

sorted_f = sorted(followers)
n = len(sorted_f)
if n % 2 == 0:
    median_val = (sorted_f[n//2 - 1] + sorted_f[n//2]) / 2
else:
    median_val = sorted_f[n//2]

print("Mean:", mean_val)
print("Median:", median_val)
```

```python
# The mean (20.125) is much higher than the median (9.5). The mean is being
# pulled up by the one large value (95), so this data is right skewed.
```

```python
#Task 3: Real world example of right skewed data
print("""
Example: YouTube video views on a channel.

Most videos on a channel usually get a similar, moderate number of views,
but once in a while one video goes viral and gets millions of views while
the rest stay in the thousands. That one (or few) viral video pulls the
average up and creates a long tail on the right side, so the data is
right skewed.
""")
```

```python
#Task 4: Distribution of Flipkart product ratings
ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]

rating_counts = {}
for r in ratings:
    rating_counts[r] = rating_counts.get(r, 0) + 1

plt.bar(rating_counts.keys(), rating_counts.values())
plt.xlabel("Rating")
plt.ylabel("Count")
plt.title("Rating Distribution")
plt.show()
```

```python
# Most ratings are on the higher side (4 and 5), with only a few low
# ratings (3). This creates a longer tail on the left, so the distribution
# is left skewed.
```
