```python
#Task 2: 3x3 matrix of friend ratings for Swiggy food items
ratings = [
    [4, 5, 3],   # Friend 1's ratings for item 1, item 2, item 3
    [3, 4, 5],   # Friend 2's ratings
    [5, 2, 4],   # Friend 3's ratings
]

print("Ratings matrix:")
for row in ratings:
    print(row)

print("\nRating each friend gave to the second food item:")
for i in range(len(ratings)):
    print("Friend", i + 1, "->", ratings[i][1])
```

```python
#Task 3: Total orders for each dish across 3 days
# rows = days, columns = dishes (4 dishes)
orders = [
    [10, 5, 8, 12],   # Day 1
    [7, 6, 9, 11],    # Day 2
    [9, 8, 10, 13],   # Day 3
]

num_dishes = len(orders[0])
dish_totals = [0] * num_dishes

for day in orders:
    for i in range(num_dishes):
        dish_totals[i] += day[i]

for i in range(num_dishes):
    print("Dish", i + 1, "total orders:", dish_totals[i])
```

```python
#Task 4: Revenue for each Flipkart product (price x quantity)
# each row: [price, quantity_sold]
products = [
    [500, 10],
    [1200, 4],
    [300, 20],
    [8000, 2],
    [150, 50],
]

def calculate_revenue(products):
    revenues = []
    for price, quantity in products:
        revenues.append(price * quantity)
    return revenues


revenues = calculate_revenue(products)

for i in range(len(products)):
    print("Product", i + 1, "revenue:", revenues[i])
```
