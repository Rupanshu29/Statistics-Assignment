```python
#Task 2: Simulate random sampling of user IDs
import random

user_ids = ["user" + str(i) for i in range(1, 1001)]

sample = random.sample(user_ids, 50)

print("Selected sample of 50 users:")
print(sample)
```

```python
#Task 3: Why use sampling for Swiggy delivery times
print("""
Collecting delivery time data from every single order ever made would take
too much time and computing power, especially since Swiggy has millions of
orders. Instead, taking a random sample of orders (say a few thousand) gives
a good estimate of the average delivery time much faster and cheaper, while
still being fairly accurate.
""")
```

```python
#Task 4: Bias in the music preference survey
print("""
This introduces sampling bias, specifically selection bias. The survey is
only sent to users who recently streamed Bollywood songs, so the sample is
not representative of all Spotify users. It will show Bollywood music as
much more popular than it might actually be for the full user base, since
users who prefer other genres (like Pop, Rock, etc.) are left out completely.
""")
```

```python
#Task 5: Two sampling types for BookMyShow movie booking habits
print("""
1. Random sampling
   Take the full list of all BookMyShow users and use random.sample() to
   pick, say, 500 users at random to study their booking habits. Every
   user has an equal chance of being picked.

2. Stratified sampling
   Split users into groups (strata) based on city or age group, then pick
   a proportional number of users randomly from each group. This makes
   sure smaller cities or age groups are not left out, and the sample
   better represents all types of users.
""")
```
