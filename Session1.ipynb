### Task 2: Classify the music genre data

Data: ['Pop', 'Rock', 'Jazz', 'Hip-Hop', 'Pop', 'Rock', 'Jazz', 'Pop']

This is categorical data because each value is a genre name, not a number.
It is also nominal, since there is no natural order between Pop, Rock, Jazz and Hip-Hop
(one genre is not "higher" or "lower" than another, they are just different labels).

### Task 3: Feature and how statistics can help

Feature: number of likes on an Instagram post.

Statistics can help by finding the average number of likes a user usually gets,
so the app can show if a new post is doing better or worse than usual.
It can also help detect outlier posts (like one that suddenly goes viral) and
use that data to recommend the best time to post or suggest similar content to more people.

### Task 4: Statistical terms for playlist recommendation

1. Mean - average listening time per song. Example: if a user listens to songs for 180, 200 and 160 seconds, the mean is 180 seconds.
2. Median - the middle value of songs played per day, useful when a few days have unusually high or low listening (like a long weekend binge). Example: if daily songs played are 10, 12, 50, the median (12) shows a more typical day than the mean.
3. Mode - the most frequently played genre or song by a user. Example: if a user plays Pop most often, Pop is the mode and can be used to recommend more Pop playlists.
