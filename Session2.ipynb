```python
#Task 2: Median of Zomato delivery times
times = [30, 25, 40, 35, 30, 45, 30]

times.sort()
n = len(times)

if n % 2 == 0:
    median = (times[n//2 - 1] + times[n//2]) / 2
else:
    median = times[n//2]

print("Sorted times:", times)
print("Median delivery time:", median)
```

```python
#Task 3: Mode of Flipkart ratings
def most_common_rating(ratings):
    count = {}
    for r in ratings:
        if r in count:
            count[r] += 1
        else:
            count[r] = 1

    best_rating = ratings[0]
    best_count = 0
    for r in count:
        if count[r] > best_count:
            best_count = count[r]
            best_rating = r
    return best_rating


ratings = [5, 4, 4, 3, 5, 4, 2, 4]
print("Ratings:", ratings)
print("Most common rating:", most_common_rating(ratings))
```

```python
#Task 4: Compare mean, median and mode for 3 channels
import statistics

channel_a = [100, 120, 110, 105, 115]
channel_b = [80, 200, 90, 85, 95]
channel_c = [100, 120, 110, 105, 5000]

for name, views in [("Channel A", channel_a), ("Channel B", channel_b), ("Channel C", channel_c)]:
    mean_v = statistics.mean(views)
    median_v = statistics.median(views)
    mode_v = statistics.mode(views)
    print(name, "-> mean:", mean_v, " median:", median_v, " mode:", mode_v)
```

```python
#Task 5: Mean, median, mode on a real dataset (IPL player runs)
import pandas as pd

df = pd.read_csv("ipl_players.csv")
df
```
