```python
#Task 1: Null and alternative hypothesis for Instagram vs YouTube time
print("""
H0 (null hypothesis): People spend the same amount of time (or less) on
Instagram compared to YouTube daily. There is no real difference.

H1 (alternative hypothesis): People spend more time on Instagram than on
YouTube daily.
""")
```

```python
#Task 2: Two-sample t-test for Zomato home page A/B test
from scipy import stats

mean_a = 350
mean_b = 320
std_dev = 50
n_a = 30
n_b = 30

t_stat, p_value = stats.ttest_ind_from_stats(
    mean1=mean_a, std1=std_dev, nobs1=n_a,
    mean2=mean_b, std2=std_dev, nobs2=n_b
)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("p-value is less than 0.05, so we reject the null hypothesis.")
else:
    print("p-value is greater than 0.05, so we fail to reject the null hypothesis.")
```

```python
# Since the p-value (about 0.024) is less than 0.05, we reject the null
# hypothesis. This means the new Zomato home page (Group A) likely does lead
# to a higher average order value than the old home page (Group B).
```
