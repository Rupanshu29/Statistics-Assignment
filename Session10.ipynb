```python
#Task 1: Function to calculate posterior probability using Bayes theorem
def bayes_posterior(prior, likelihood, evidence):
    return (prior * likelihood) / evidence


# Example values
posterior = bayes_posterior(0.3, 0.7, 0.5)
print("Posterior probability:", posterior)
```

```python
#Task 2: Probability a Swiggy restaurant is premium given it offers a discount
prior = 0.10                  
likelihood = 0.80             
p_discount_given_regular = 0.20   

evidence = (likelihood * prior) + (p_discount_given_regular * (1 - prior))

posterior = bayes_posterior(prior, likelihood, evidence)
print("P(discount) - evidence:", evidence)
print("P(premium | discount):", posterior)
```

```python
# So if you see a restaurant offering a discount, there is about a 30.8%
# chance it is actually a premium restaurant.
```

```python
#Task 3: Spam email filter using Bayes theorem
emails = [
    ("Get your free gift now", "spam"),
    ("Meeting rescheduled to 3pm", "not spam"),
    ("Free entry to win a prize", "spam"),
    ("Your invoice for last month", "not spam"),
    ("Claim your free reward today", "spam"),
    ("Project update from the team", "not spam"),
    ("You have won a free vacation", "spam"),
    ("Lunch plans for tomorrow", "not spam"),
    ("Free trial ending soon, act now", "spam"),
    ("Notes from today's class", "not spam"),
    ("Limited offer, click for free stuff", "spam"),
    ("Reminder for tomorrow's assignment", "not spam"),
    ("Congratulations, you get a free coupon", "spam"),
    ("Weekly newsletter from college", "not spam"),
    ("Free money waiting for you", "spam"),
    ("Group project discussion", "not spam"),
    ("Your free sample has shipped", "spam"),
    ("Fees payment confirmation", "not spam"),
    ("Act now for a free upgrade", "spam"),
    ("Free shipping on your recent order", "not spam"),
]

spam_emails = [subject for subject, label in emails if label == "spam"]
not_spam_emails = [subject for subject, label in emails if label == "not spam"]

spam_with_free = sum(1 for s in spam_emails if "free" in s.lower())
not_spam_with_free = sum(1 for s in not_spam_emails if "free" in s.lower())

prior_spam = len(spam_emails) / len(emails)
likelihood_free_given_spam = spam_with_free / len(spam_emails)
likelihood_free_given_not_spam = not_spam_with_free / len(not_spam_emails)

evidence_free = (likelihood_free_given_spam * prior_spam) + \
                (likelihood_free_given_not_spam * (1 - prior_spam))

posterior_spam_given_free = bayes_posterior(prior_spam, likelihood_free_given_spam, evidence_free)

print("P(spam):", prior_spam)
print("P('free' | spam):", likelihood_free_given_spam)
print("P('free' | not spam):", likelihood_free_given_not_spam)
print("P(spam | 'free'):", posterior_spam_given_free)
```

```python
#Task 4: Real world Bayes theorem example (Instagram)
print("""
Prompt used: "Generate a real world Bayes theorem example related to
Instagram, involving probability a post goes viral given it uses a
trending hashtag."

Example generated: On Instagram, 5% of all posts go viral. Among viral
posts, 60% used a trending hashtag. Among non-viral posts, only 15% used a
trending hashtag. If a post uses a trending hashtag, what is the
probability it goes viral?
""")

prior_viral = 0.05
likelihood_hashtag_given_viral = 0.60
p_hashtag_given_not_viral = 0.15

evidence_hashtag = (likelihood_hashtag_given_viral * prior_viral) + \
                    (p_hashtag_given_not_viral * (1 - prior_viral))

posterior_viral_given_hashtag = bayes_posterior(prior_viral, likelihood_hashtag_given_viral, evidence_hashtag)

print("P(viral | trending hashtag):", posterior_viral_given_hashtag)
```
