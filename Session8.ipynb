```python
#Task 1: Function to calculate probability
def calculate_probability(event_count, total_outcomes):
    return event_count / total_outcomes


# Probability of getting Heads when tossing a coin
heads_probability = calculate_probability(1, 2)
print("Probability of Heads:", heads_probability)
```

```python
#Task 2: Simulate rolling a dice 100 times and find probability of getting a 6
import random

rolls = 100
six_count = 0

for i in range(rolls):
    result = random.randint(1, 6)
    if result == 6:
        six_count += 1

probability_of_six = six_count / rolls
print("Number of times 6 appeared:", six_count)
print("Experimental probability of getting a 6:", probability_of_six)
```

```python
#Task 3: Probability of picking a Pop song from recent history
genres = ['pop', 'rock', 'pop', 'hiphop', 'pop', 'rock', 'jazz', 'pop', 'hiphop', 'pop',
          'rock', 'pop', 'jazz', 'hiphop', 'pop', 'rock', 'pop', 'jazz', 'hiphop', 'pop']

pop_count = 0
for g in genres:
    if g == 'pop':
        pop_count += 1

probability_pop = pop_count / len(genres)
print("Genres:", genres)
print("Number of Pop songs:", pop_count)
print("Probability of picking a Pop song:", probability_pop)
```

```python
#Task 4: Probability a random user ordered both food and dessert
def probability_both(food_users, dessert_users):
    food_set = set(food_users)
    dessert_set = set(dessert_users)

    both = food_set.intersection(dessert_set)
    total_unique_users = food_set.union(dessert_set)

    return len(both) / len(total_unique_users)


zomato_users = ["u1", "u2", "u3", "u4", "u5"]
dessert_users = ["u3", "u4", "u6", "u7"]

print("Probability a random user ordered both food and dessert:",
      probability_both(zomato_users, dessert_users))
```

```python
#Task 5: Conditional probability example for Flipkart recommendations
phone_buyers = ["u1", "u2", "u3", "u4", "u5", "u6", "u7", "u8"]
headphone_buyers = ["u2", "u3", "u5", "u8"]

both_count = 0
for user in headphone_buyers:
    if user in phone_buyers:
        both_count += 1

# P(headphones | phone) = users who bought both / users who bought a phone
probability_headphones_given_phone = both_count / len(phone_buyers)

print("P(headphones | phone):", probability_headphones_given_phone)
# This tells Flipkart that a good chunk of phone buyers also buy headphones,
# so it makes sense to recommend headphones to anyone who just bought a phone.
```
