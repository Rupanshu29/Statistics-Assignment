```python
#Task 1: Variance of daily step counts (by hand, step by step)
steps = [4500, 7000, 5000, 8000, 4000, 9000, 6000]

n = len(steps)
total = sum(steps)
mean = total / n
print("Step 1: Mean = sum / n =", total, "/", n, "=", mean)

squared_diffs = []
for x in steps:
    diff = x - mean
    sq = diff ** 2
    squared_diffs.append(sq)
    print(f"({x} - {mean}) squared = {sq}")

sum_sq = sum(squared_diffs)
print("\nStep 2: Sum of squared differences =", sum_sq)

variance = sum_sq / n
print("Step 3: Variance = sum_sq / n =", sum_sq, "/", n, "=", variance)
```

```python
#Task 2: Function to calculate standard deviation
def calculate_standard_deviation(scores):
    n = len(scores)
    mean = sum(scores) / n

    squared_diffs = []
    for x in scores:
        squared_diffs.append((x - mean) ** 2)

    variance = sum(squared_diffs) / n
    std_dev = variance ** 0.5
    return round(std_dev, 2)


exam_scores = [78, 85, 92, 66, 74, 90]
print("Standard deviation:", calculate_standard_deviation(exam_scores))
```

```python
#Task 3: Compare spending consistency of two friends
friend_a = [200, 200, 200, 200, 200]
friend_b = [100, 300, 150, 400, 50]

std_a = calculate_standard_deviation(friend_a)
std_b = calculate_standard_deviation(friend_b)

print("Friend A standard deviation:", std_a)
print("Friend B standard deviation:", std_b)
```

```python
#Task 4: Variance and standard deviation of employee salaries
salaries = [28000, 29000, 27000, 60000, 26500, 27500]

variance_salary = sum((x - sum(salaries)/len(salaries)) ** 2 for x in salaries) / len(salaries)
std_salary = variance_salary ** 0.5

print("Variance:", variance_salary)
print("Standard deviation:", round(std_salary, 2))
```
