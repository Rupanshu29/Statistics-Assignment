```python
#Task 1: Generate normal distribution data and plot the bell curve
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 100)

plt.hist(data, bins=15)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Normal Distribution (mean=50, std=10)")
plt.show()
```

```python
#Task 2: Mean, median and mode of daily step counts
from scipy import stats
import numpy as np

steps = np.array([5000, 5200, 5100, 5000, 4900, 5300, 5100, 5000, 5200, 5100,
                   4900, 5000, 5100, 5200, 5000, 4900, 5100, 5000, 5200, 5100])

mean_steps = np.mean(steps)
median_steps = np.median(steps)
mode_steps = stats.mode(steps, keepdims=True)

print("Mean:", mean_steps)
print("Median:", median_steps)
print("Mode:", mode_steps.mode[0])
```

```python
# The mean, median and mode all come out close to 5000-5100, so they are
# approximately equal. This matches what we would expect from data that
# follows a normal (bell shaped) distribution.
```

```python
#Task 3: Simulate 200 player scores and apply the 68-95-99 rule
import numpy as np

scores = np.random.normal(70, 15, 200)

within_one_std = 0
for s in scores:
    if 55 <= s <= 85:
        within_one_std += 1

print("Players within one standard deviation (55 to 85):", within_one_std)
print("Percentage:", (within_one_std / 200) * 100)
```

```python
# According to the 68-95-99 rule, about 68% of values should fall within
# one standard deviation of the mean. Our simulated result is close to that,
# which confirms the data behaves like a normal distribution.
```
